# Step 2: Data Cleaning and Preprocessing

- Identify Missing values

```python
missing_val=df.isnull().sum()
print(missing_val[missing_val > 0])
```

### Missing data treatment

```python
# Missing data treatment for categorical data(Customer Lname)
df['Customer Lname']=df['Customer Lname'].fillna('Unknown')
```

```python
df['Customer Zipcode']=(df['Customer Zipcode'].astype('string').fillna('Unknown'))
```

```python
# Drop columns with excessive missing values
df.drop(columns=['Order Zipcode', 'Product Description'],inplace=True)
```

```python
print('Missing values after cleaning:\n',df.isnull().sum().sum())
```

### Check and handle duplicate records

```python
duplicates = df.duplicated().sum()
print("Total Duplicate Rows:", duplicates)
```

- **Finding:** No duplicate records found.

### Date Formatting

```python
df['order date (DateOrders)']=pd.to_datetime(df['order date (DateOrders)'])
df['shipping date (DateOrders)']=pd.to_datetime(df['shipping date (DateOrders)'])
```

```python
df['shipping date (DateOrders)'].dtypes
```

```python
df['order date (DateOrders)'].dtypes
```

### Extracting year, month name, quarter and day from Order Date

```python
df['Order Year']=(df['order date (DateOrders)'].dt.year)
df['Order Month'] = (df['order date (DateOrders)'].dt.month)
df['Order Month Name'] = (df['order date (DateOrders)'].dt.month_name())
df['Order Quarter'] = (df['order date (DateOrders)'].dt.quarter)
df['Order Day'] = (df['order date (DateOrders)'].dt.day_name())
```

```python
print("DATA CLEANING & PREPROCESSING SUMMARY")
print("=" * 55)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Remaining Missing Values:",df.isnull().sum().sum())
print("Remaining Duplicate Rows:",df.duplicated().sum())
print("=" * 55)
```


### Key Findings: 

1.  Missing values identified.
2.  Missing values in `Customer Lname` (8) and `Customer Zipcode` (3) were replaced with `"Unknown"`.
3. `Order Zipcode` (86.24% missing) and `Product Description` (100% missing) were removed due to insufficient data.
4.  Exact duplicate rows were identified .
5.  Order and shipping dates were converted into datetime format.
6.  Important  attributes were checked and converted where necessary.

# Step 3: Exploratory Data Analysis (EDA)
### ***Objective***- 

The objective of Exploratory Data Analysis is to identify patterns, trends, and relationships within the supply chain dataset.

- This analysis focuses on:

1. Overall sales and profitability performance
2. Monthly and yearly sales trends
3. Top-performing product categories
4. Customer segment performance
5. Market-wise and regional sales distribution
6. Order status and fulfillment analysis
7. Shipping and delivery performance

- Import Visualization Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Q.1 How has sales performance changed over the years?

```python
yearly_sales = (df.groupby('Order Year')['Sales'].sum().reset_index())
yearly_sales
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=yearly_sales,
    x='Order Year',
    y='Sales'
)
plt.title('Yearly Sales Performance', fontsize=15)
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Observation:
- The bar chart shows that 2015 recorded the highest total sales (approximately 12.3 million), followed by 2016 (12.1 million) and 2017 (11.5 million), indicating a gradual decline in annual sales. Sales in 2018 dropped sharply to approximately 0.33 million, primarily because the dataset contains only January 2018 records.
### Conclusion:
- The analysis indicates a slight downward trend in yearly sales from 2015 to 2017, suggesting the need to investigate declining revenue performance.
- The significant drop in 2018 reflects incomplete yearly data rather than a confirmed business decline.
-  Businesses should focus on improving sales strategies, customer retention, and revenue growth.

### Q2. Which product categories generate the highest sales?

```python
category_sales =(df.groupby('Category Name')['Sales'].sum().sort_values(ascending=False)).head(10)
plt.figure(figsize=(11, 6))
sns.barplot(
    x=category_sales.values,
    y=category_sales.index,
    hue=category_sales.index,
    legend=False
)
plt.title('Top 10 Product Categories by Sales', fontsize=15)
plt.xlabel('Total Sales')
plt.ylabel('Product Category')
plt.show()
```

### Observation: 
- The chart identifies the ten highest-revenue product categories and shows the difference in sales contribution between them.
- The category ranked first represents the largest revenue contributor.

### Conclusion: 
- The highest-revenue categories are important for sales planning and product prioritization.The Fishing category has highest sales.

### Q.3 How is sales revenue distributed across customer segments?

```python
segment_sales = (df.groupby('Customer Segment')['Sales'].sum().sort_values(ascending=False))
plt.figure(figsize=(10, 5))
plt.pie(
    segment_sales,
    labels=segment_sales.index,
    autopct='%1.1f%%',
    startangle=90,
    shadow= True
)
plt.title('Sales Contribution by Customer Segment', fontsize=15)
plt.show()
```

### Observation:
- The pie chart represents the percentage contribution of different customer segments to overall sales revenue.
  
### Conclusion:
- The highest-contributing customer segment represents the largest source of sales revenue and should be prioritized for customer retention.

### Q.4 What is the relationship between sales and profit?

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(
    data=df,
    x='Sales',
    y='Order Item Profit Ratio',
    alpha=0.4
)
plt.title('Relationship Between Sales and Profit Ratio', fontsize=15)
plt.xlabel('Sales Amount')
plt.ylabel('Profit Ratio')
plt.show()
```

### Observation:
- The scatter plot illustrates the relationship between Sales Amount and Order Item Profit Ratio, showing how profitability varies across different sales values.
### Conclusion:
- The analysis highlights that higher sales revenue does not necessarily guarantee higher profitability, as profitability depends on the profit ratio associated with individual transactions.
- Orders with negative profit ratios indicate loss-making transactions that require further investigation into pricing, discounts, and operational costs.

### Q.5 What is the distribution of shipping delays?

```python
df['Shipping Delay Days'] = (df['Days for shipping (real)'] -df['Days for shipment (scheduled)'])
plt.figure(figsize=(10,5))
sns.histplot(
    data=df,
    x='Shipping Delay Days',
    discrete=True,
    color='steelblue'
)
plt.title('Distribution of Shipping Delays', fontsize=15)
plt.xlabel('Shipping Delay (Days)')
plt.ylabel('Number of Records')
plt.show()
```

### Observation:
- The Shipping Delay Days column measures the difference between actual and scheduled shipping durations. Positive values indicate delayed shipments, negative values indicate early shipments, and zero indicates on-time shipping.
### Conclusion:
-  Reducing positive delay values can enhance operational efficiency and customer satisfaction.

### Q.6 How does profitability vary across markets?

```python
plt.figure(figsize=(10, 5))
sns.boxplot(
    data=df,
    x='Market',
    y='Order Item Profit Ratio',
    hue='Market',
    showfliers=False
)

plt.axhline(0, color='red', linestyle='--')

plt.title('Profit Ratio Distribution Across Markets', fontsize=15)
plt.xlabel('Market')
plt.ylabel('Order Item Profit Ratio')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Observation:
- The box plot shows that Pacific Asia has the highest median profit ratio (approximately 0.26), followed by USCA (0.25), Africa (0.23), Europe (0.22), and LATAM (0.20). All markets have positive median profit ratios, but their distributions extend below zero, indicating the presence of loss-making transactions across every market.
### Conclusion:
- Pacific Asia demonstrates the strongest typical profitability, while LATAM has the lowest median profit ratio.

### Q.7  How do monthly sales patterns vary across years?

```python
monthly_sales = df.pivot_table(
    index='Order Year',
    columns='Order Month',
    values='Sales',
    aggfunc='sum'
)
monthly_sales = monthly_sales.reindex(columns=range(1, 13))
monthly_sales.columns = [ 'Jan', 'Feb', 'Mar', 'Apr','May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.figure(figsize=(13, 6))
sns.heatmap(
    monthly_sales,
    annot=True,
    fmt='.0f',
    linewidths=0.5
)
plt.title('Monthly Sales Performance Across Years', fontsize=15)
plt.xlabel('Month')
plt.ylabel('Year')
plt.tight_layout()
plt.show()
```

### Observation:
- The heatmap shows that monthly sales remained relatively stable during 2015 and 2016, averaging around 1 million per month. 
- In 2017, sales peaked in September (1.14 million) before declining sharply in November (0.63 million) and December (0.50 million).
- January 2018 recorded the lowest sales (0.33 million), with no data available for the remaining months.
### Conclusion:
- The analysis indicates consistent sales performance during 2015–2016, followed by fluctuations in 2017. The sharp decline toward the end of 2017 requires further investigation into seasonal demand, business operations, or incomplete data.
- Businesses should monitor monthly sales trends to improve demand forecasting, inventory planning, and revenue stability.
